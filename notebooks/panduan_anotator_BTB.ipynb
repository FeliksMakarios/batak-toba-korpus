# **Batak Toba Textual Corpus Project**

## **1. What "Text" Means in Corpus Linguistics**

In corpus linguistics, a **"text"** refers to a **complete document or coherent unit of discourse**, not individual sentences. Examples include:

- **Historical/Traditional**: Complete manuscript, full genealogy, entire folktale
- **Literary**: Complete short story, full poem, novel chapter  
- **Religious**: Complete prayer, full sermon, ritual instruction set
- **Educational**: Textbook chapter, complete lesson, full primer
- **Contemporary Media**: Complete news article, full blog post, entire magazine article

### 2.1 Text Types and Genres
| **Genre Category** | **Text Types** | **Rationale** | **Target %** |
|---|---|---|---|
| Historical/Traditional | Manuscripts (pustaha laklak), folklore, genealogies | Cultural preservation, historical linguistics | 25% |
| Literary | Novels, poetry, short stories | Stylistic diversity, modern usage | 20% |
| Religious | Prayer books, sermons, ritual texts | Domain representation, community significance | 15% |
| Educational | Textbooks, teaching materials, primers | Language standardization, pedagogical use | 15% |
| Contemporary Media | Newspapers, magazines, blogs | Current usage patterns, sociolinguistic variation | 15% |
| Administrative | Official documents, letters (public domain) | Formal register, institutional language | 10% |
